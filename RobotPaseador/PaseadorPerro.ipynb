```python
"""
Robot Paseador de Perros - Simulación en Python para Google Colab
Basado en la descomposición: Tarea Principal -> 5 Subtareas -> Pasos específicos (P/S)

Como el robot no existe físicamente, los sensores se SIMULAN con la librería
`random`. Cada INPUT del pseudocódigo se convierte en una lectura simulada,
y cada OUTPUT se convierte en un print() que representa la acción/comunicación
del robot.
"""

import random
import time

# =========================================================
# ESTADO GLOBAL DEL ROBOT
# Guarda los datos que varias funciones necesitan compartir
# =========================================================
estado = {
    "tiempo_transcurrido": 0,          # minutos transcurridos desde que empezó el paseo
    "tiempo_limite_regreso": 0,
    "ruta": [],
    "notificaciones_activas": True,
    "bolsita_recogida": False,
    "distancia_total": 0,
    "paradas_realizadas": 0,
    "incidencias": [],
    "robot_en_casa": False,
}


# =========================================================
# SUBTAREA 1: PREPARAR
# =========================================================
def preparar():
    print("\n--- SUBTAREA 1: Preparar el paseo ---")

    # Paso 1 (P) Revisión inicial
    nivel_bateria = random.randint(0, 100)       # INPUT simulado: sensor de batería
    correa_conectada = True                       # INPUT simulado: sensor de correa
    perro_identificado_correcto = True                       # INPUT simulado: lector RFID

    print(f"INPUT nivel_bateria = {nivel_bateria}%")

    if nivel_bateria < 20:
        print("OUTPUT: Alerta -> Batería baja, cargar antes de salir")
        return False  # detiene el programa

    if not correa_conectada or not perro_identificado_correcto:
        print("OUTPUT: Alerta -> Revisión inicial fallida")
        return False

    # Paso 2 (P) Planificar la ruta
    duracion_planeada = 30  # INPUT: minutos, definido por el dueño (o valor por defecto)
    estado["tiempo_limite_regreso"] = duracion_planeada
    estado["ruta"] = ["calle_1", "calle_2", "parque", "calle_3"]
    print(f"OUTPUT: Ruta planificada = {estado['ruta']}, regreso en {duracion_planeada} min")

    # Paso 3 (P) Salir de casa
    print("OUTPUT: Puerta abierta -> el robot sale de casa")

    # Paso 4 (S) Avisar al dueño
    if estado["notificaciones_activas"]:
        print("OUTPUT: Notificación al dueño -> 'El paseo ha comenzado'")

    return True


# =========================================================
# SUBTAREA 2: CAMINAR LA RUTA
# =========================================================
def caminar_ruta():
    print("\n--- SUBTAREA 2: Caminar la ruta ---")

    # Paso 1 (P) Caminar al ritmo del perro
    velocidad_perro = random.choice(["lenta", "normal", "rapida"])  # INPUT simulado
    print(f"INPUT velocidad_perro = {velocidad_perro}")
    print(f"OUTPUT: El robot ajusta su velocidad a '{velocidad_perro}'")

    # Paso 2 (P) Esquivar charcos y obstáculos
    distancia_obstaculo = random.randint(0, 5)     # INPUT simulado: sensor ultrasónico
    superficie_detectada = random.choice(["seca", "charco"])  # INPUT simulado: sensor de humedad
    print(f"INPUT distancia_obstaculo = {distancia_obstaculo}m, superficie_detectada = {superficie_detectada}")

    if distancia_obstaculo < 1 or superficie_detectada == "charco":
        print("OUTPUT: El robot esquiva el obstáculo/charco")
        estado["incidencias"].append("esquivó obstáculo")

    estado["distancia_total"] += 1  # avanza 1 unidad de ruta

    # Paso 3 (P) Controlar el tiempo restante
    # `tiempo_transcurrido` es la VARIABLE DE CONTROL del paseo: es la que el
    # bucle principal (en main()) revisa en cada vuelta para decidir si sigue
    # caminando o si ya debe finalizar. Su propósito es exactamente ese control,
    # no solo informar la hora.
    estado["tiempo_transcurrido"] += 5  # avanza 5 minutos por ciclo
    minutos_disponibles = estado["tiempo_limite_regreso"] - estado["tiempo_transcurrido"]
    print(f"INPUT tiempo_transcurrido = {estado['tiempo_transcurrido']} min, minutos_disponibles = {minutos_disponibles} min")

    if minutos_disponibles < 5:
        print("OUTPUT: Tiempo ajustado -> se acorta la ruta")
        estado["incidencias"].append("ruta acortada por tiempo")


# =========================================================
# SUBTAREA 3: REACCIONAR A IMPREVISTOS
# =========================================================
def reaccionar_imprevistos():
    print("\n--- SUBTAREA 3: Reaccionar a imprevistos ---")

    # Paso 1 (S) Evitar ruidos fuertes
    nivel_ruido = random.randint(0, 100)  # INPUT simulado: micrófono
    print(f"INPUT nivel_ruido = {nivel_ruido} dB")
    if nivel_ruido > 80:
        print("OUTPUT: Ruta actualizada -> cambia de calle por ruido fuerte")
        estado["incidencias"].append("cambió de calle por ruido")

    # Paso 2 (S) Cambiar de ruta si llueve
    esta_lloviendo = random.choice([True, False])  # INPUT simulado: sensor de lluvia
    print(f"INPUT esta_lloviendo = {esta_lloviendo}")
    if esta_lloviendo:
        print("OUTPUT: Ruta actualizada -> toma el camino más corto por lluvia")
        estado["incidencias"].append("acortó ruta por lluvia")


# =========================================================
# SUBTAREA 4: ATENDER AL PERRO
# =========================================================
def atender_perro():
    print("\n--- SUBTAREA 4: Atender al perro ---")

    # Paso 1 (P) Recoger la bolsita
    necesita_recoger_bolsita = random.choice([True, False])  # INPUT simulado: cámara/sensor
    print(f"INPUT necesita_recoger_bolsita = {necesita_recoger_bolsita}")

    if necesita_recoger_bolsita:
        print("OUTPUT: El robot se detiene y activa el brazo recolector")
        estado["bolsita_recogida"] = True
        estado["paradas_realizadas"] += 1

        # Paso 2 (S) Tirar la bolsita
        ubicacion_papelera = "papelera_cercana"  # INPUT simulado: mapa/GPS
        print(f"INPUT ubicacion_papelera_cercana = {ubicacion_papelera}")
        print(f"OUTPUT: Bolsita depositada en {ubicacion_papelera}")
        estado["bolsita_recogida"] = False

    # Paso 3 (S) Ofrecer agua
    temperatura_ambiente = random.randint(15, 35)   # INPUT simulado: sensor de temperatura
    nivel_cansancio = random.choice(["bajo", "medio", "alto"])  # INPUT simulado
    print(f"INPUT temperatura_ambiente = {temperatura_ambiente}°C, cansancio = {nivel_cansancio}")

    if temperatura_ambiente > 28 or nivel_cansancio == "alto":
        print("OUTPUT: El robot se detiene y dispensa agua")
        estado["paradas_realizadas"] += 1


# =========================================================
# SUBTAREA 5: FINALIZAR EL PASEO
# =========================================================
def finalizar():
    print("\n--- SUBTAREA 5: Finalizar el paseo ---")

    # Paso 1 (P) Regresar a casa
    print("OUTPUT: El robot calcula la ruta más corta y regresa a casa")

    # Paso 2 (P) Entregar al perro
    llegada_confirmada = True  # INPUT simulado: geolocalización/sensor de puerta
    print(f"INPUT llegada_confirmada = {llegada_confirmada}")
    if llegada_confirmada:
        print("OUTPUT: Correa liberada -> el perro es entregado en casa")
        estado["robot_en_casa"] = True

    # Paso 3 (S) Guardar registro
    registro = {
        "distancia_total": estado["distancia_total"],
        "duracion_total": estado["tiempo_transcurrido"],
        "paradas_realizadas": estado["paradas_realizadas"],
        "incidencias": estado["incidencias"],
    }
    print(f"OUTPUT: Registro guardado -> {registro}")

    # Paso 4 (S) Avisar llegada
    if estado["notificaciones_activas"]:
        print("OUTPUT: Notificación al dueño -> 'El perro ya está en casa'")


# =========================================================
# PROGRAMA PRINCIPAL
# =========================================================
def main():
    if not preparar():
        return  # se detiene si la revisión inicial falla

    # CONDICIÓN DE CONTROL DEL BUCLE:
    # mientras tiempo_transcurrido no alcance tiempo_limite_regreso Y el robot
    # no haya llegado a casa, se repite el ciclo caminar -> reaccionar -> atender.
    # Estas dos variables ("tiempo_transcurrido" y "robot_en_casa") son las que
    # controlan cuántas vueltas da el paseo antes de finalizar.
    while estado["tiempo_transcurrido"] < estado["tiempo_limite_regreso"] and not estado["robot_en_casa"]:
        caminar_ruta()
        reaccionar_imprevistos()
        atender_perro()
        time.sleep(0.3)  # solo para que la simulación se lea con pausas en Colab

    finalizar()


if __name__ == "__main__":
    main()


```

```python
""" MERGE MIO y GRUPO
Robot Paseador de Perros
=========================
Implementación en Python del algoritmo definido en pseudocódigo.

Estructura (Nivel 1 - Pasos principales):
    1. Preparar el paseo
    2. Caminar la ruta
    3. Atender al perro
    4. Resolver imprevistos (transversal a 2 y 3)
    5. Finalizar el paseo

Nota: como no hay hardware real conectado, las funciones de "sensores"
(detectar_lluvia, detectar_heces, etc.) están SIMULADAS con probabilidades
aleatorias, para que el flujo completo sea ejecutable y demostrable.
Sustituye esas funciones por las lecturas reales de sensores/cámaras cuando
se integre con el robot físico.
"""

import random
import time
from dataclasses import dataclass, field
from datetime import datetime, timedelta
from enum import Enum


# ============================================================
# CONFIGURACIÓN / CONSTANTES
# ============================================================

BATERIA_MINIMA = 20          # % mínimo para poder salir
BATERIA_CRITICA = 10         # % que fuerza el regreso inmediato
DURACION_MAXIMA_PASEO = timedelta(minutes=30)
MARGEN_SEGURIDAD_REGRESO = timedelta(minutes=2)  # colchón extra al calcular si hay que volver


class Imprevisto(Enum):
    NINGUNO = "NINGUNO"
    BATERIA_BAJA = "BATERIA_BAJA"
    LLUVIA = "LLUVIA"
    OBSTACULO = "OBSTACULO"
    RUIDO = "RUIDO"
    SIN_RUTA_ALTERNATIVA = "SIN_RUTA_ALTERNATIVA"


# ============================================================
# ESTADO DEL ROBOT
# ============================================================

@dataclass
class EstadoRobot:
    bateria: float = 100.0
    tiene_correa: bool = True
    tiene_bolsas: bool = True
    tiene_agua: bool = True

    ruta: list = field(default_factory=list)
    posicion_actual: int = 0          # índice dentro de la ruta
    en_casa: bool = True

    hora_inicio: datetime = None
    hora_limite_regreso: datetime = None

    bolsita_pendiente: bool = False   # True si recogió una bolsita y no encontró papelera aún
    incidentes: list = field(default_factory=list)
    dueño_avisado_inicio: bool = False
    dueño_avisado_fin: bool = False


# ============================================================
# FUNCIONES "SENSOR" SIMULADAS
# (reemplazar por lecturas reales de hardware)
# ============================================================

def revisar_bateria(estado: EstadoRobot) -> float:
    return estado.bateria


def verificar_insumos(estado: EstadoRobot) -> dict:
    return {
        "correa": estado.tiene_correa,
        "bolsas": estado.tiene_bolsas,
        "agua": estado.tiene_agua,
    }


def planificar_ruta() -> list:
    """Simula una ruta como una lista de tramos/puntos a recorrer."""
    return [f"tramo_{i}" for i in range(1, 9)]


def detectar_lluvia() -> bool:
    return random.random() < 0.05


def detectar_obstaculo_o_charco() -> bool:
    return random.random() < 0.10


def detectar_ruido_fuerte() -> bool:
    return random.random() < 0.05


def hay_ruta_alternativa() -> bool:
    """Probabilidad de que exista un camino alternativo al esquivar algo."""
    return random.random() < 0.9


def detectar_heces() -> bool:
    return random.random() < 0.15


def detectar_orina() -> bool:
    return random.random() < 0.15


def perro_pide_agua() -> bool:
    return random.random() < 0.08


def hay_papelera_cerca() -> bool:
    return random.random() < 0.6


# ============================================================
# FUNCIONES "ACTUADOR" SIMULADAS
# ============================================================

def alertar(mensaje: str):
    print(f"[ALERTA] {mensaje}")


def avisar_dueño(mensaje: str):
    print(f"[NOTIFICACIÓN AL DUEÑO] {mensaje}")


def salir_de_casa(estado: EstadoRobot):
    estado.en_casa = False
    print("[ACCIÓN] Saliendo de casa...")


def recalcular_ruta(estado: EstadoRobot, motivo: str):
    print(f"[ACCIÓN] Recalculando ruta para evitar: {motivo}")


def detener_robot():
    print("[ACCIÓN] Robot detenido.")


def posicionar_brazo():
    print("[ACCIÓN] Posicionando brazo robótico...")


def tomar_bolsita_dispensador():
    print("[ACCIÓN] Bolsita tomada del dispensador.")


def recoger_residuos():
    print("[ACCIÓN] Residuos recogidos.")


def depositar_en_papelera():
    print("[ACCIÓN] Bolsita depositada en la papelera.")


def rociar_agua_limpieza():
    print("[ACCIÓN] Rociando agua para limpiar orina (según reglamento).")


def ofrecer_agua():
    print("[ACCIÓN] Ofreciendo agua para beber al perro.")


def caminar_tramo(estado: EstadoRobot):
    if estado.posicion_actual < len(estado.ruta) - 1:
        estado.posicion_actual += 1
    print(f"[ACCIÓN] Caminando... posición actual: {estado.ruta[estado.posicion_actual]}")


def calcular_tiempo_regreso(estado: EstadoRobot) -> timedelta:
    """Estimación simple: a más lejos en la ruta, más tiempo de regreso."""
    tramos_restantes_hasta_inicio = estado.posicion_actual
    minutos_por_tramo = 1.5
    return timedelta(minutes=tramos_restantes_hasta_inicio * minutos_por_tramo) + MARGEN_SEGURIDAD_REGRESO


def seguir_ruta_mas_corta_a_casa(estado: EstadoRobot):
    print("[ACCIÓN] Regresando a casa por el camino más corto...")
    estado.posicion_actual = 0
    estado.en_casa = True


def entregar_perro():
    print("[ACCIÓN] Perro entregado en casa.")


def guardar_registro(estado: EstadoRobot):
    duracion = datetime.now() - estado.hora_inicio
    print(
        f"[REGISTRO] Paseo finalizado. Duración: {duracion}. "
        f"Incidentes registrados: {estado.incidentes if estado.incidentes else 'ninguno'}."
    )


# ============================================================
# PASO 1: PREPARAR EL PASEO
# ============================================================

def preparar_paseo(estado: EstadoRobot) -> bool:
    print("\n=== PASO 1: PREPARAR EL PASEO ===")

    bateria = revisar_bateria(estado)
    if bateria < BATERIA_MINIMA:
        alertar(f"Batería insuficiente ({bateria}%), no se inicia el paseo.")
        return False

    insumos = verificar_insumos(estado)
    faltantes = [nombre for nombre, presente in insumos.items() if not presente]
    if faltantes:
        alertar(f"Falta material: {', '.join(faltantes)}. No se inicia el paseo.")
        return False

    estado.ruta = planificar_ruta()
    estado.hora_inicio = datetime.now()
    estado.hora_limite_regreso = estado.hora_inicio + DURACION_MAXIMA_PASEO

    salir_de_casa(estado)
    avisar_dueño("El paseo ha comenzado.")
    estado.dueño_avisado_inicio = True

    return True


# ============================================================
# PASO 4: RESOLVER IMPREVISTOS (evaluado antes de cada tramo)
# Orden de prioridad: batería > ruido > lluvia > obstáculo
# ============================================================

def revisar_imprevistos(estado: EstadoRobot) -> Imprevisto:
    # 1. Prioridad máxima: seguridad energética del robot
    if estado.bateria < BATERIA_CRITICA:
        estado.incidentes.append("Batería crítica: regreso forzado")
        return Imprevisto.BATERIA_BAJA

    # 2. Seguridad ante el entorno (ruido / perros ladrando)
    if detectar_ruido_fuerte():
        detener_robot()
        estado.incidentes.append("Ruido fuerte detectado: robot se detuvo")
        return Imprevisto.RUIDO

    # 3. Clima
    if detectar_lluvia():
        if hay_ruta_alternativa():
            recalcular_ruta(estado, "lluvia")
            estado.incidentes.append("Lluvia: ruta recalculada")
            return Imprevisto.LLUVIA
        else:
            estado.incidentes.append("Lluvia sin ruta alternativa: regreso forzado")
            return Imprevisto.SIN_RUTA_ALTERNATIVA

    # 4. Terreno (charcos / obstáculos)
    if detectar_obstaculo_o_charco():
        if hay_ruta_alternativa():
            recalcular_ruta(estado, "obstáculo/charco")
            estado.incidentes.append("Obstáculo/charco: ruta recalculada")
            return Imprevisto.OBSTACULO
        else:
            estado.incidentes.append("Obstáculo sin ruta alternativa: regreso forzado")
            return Imprevisto.SIN_RUTA_ALTERNATIVA

    return Imprevisto.NINGUNO


# ============================================================
# PASO 3: ATENDER AL PERRO
# ============================================================

def atender_perro(estado: EstadoRobot):
    if detectar_heces():
        detener_robot()
        posicionar_brazo()
        tomar_bolsita_dispensador()
        recoger_residuos()
        estado.bolsita_pendiente = True

    # Si tiene una bolsita pendiente (recién recogida o de un tramo anterior),
    # intenta depositarla en cuanto encuentre una papelera.
    if estado.bolsita_pendiente and hay_papelera_cerca():
        depositar_en_papelera()
        estado.bolsita_pendiente = False

    if detectar_orina():
        rociar_agua_limpieza()

    if perro_pide_agua():
        ofrecer_agua()


# ============================================================
# PASO 2: CAMINAR LA RUTA (bucle principal)
# ============================================================

def necesita_regresar_ya(estado: EstadoRobot) -> bool:
    tiempo_restante = estado.hora_limite_regreso - datetime.now()
    tiempo_necesario_regreso = calcular_tiempo_regreso(estado)
    return tiempo_restante <= tiempo_necesario_regreso


def caminar_ruta(estado: EstadoRobot):
    print("\n=== PASO 2: CAMINAR LA RUTA ===")

    while not estado.en_casa and estado.posicion_actual < len(estado.ruta) - 1:

        imprevisto = revisar_imprevistos(estado)

        if imprevisto in (Imprevisto.BATERIA_BAJA, Imprevisto.SIN_RUTA_ALTERNATIVA):
            print(f"[DECISIÓN] Imprevisto crítico ({imprevisto.value}): se interrumpe el paseo y se regresa.")
            break

        # Si hubo un imprevisto que requirió detenerse (ruido), no se avanza
        # ni se atiende al perro en esta vuelta; se reintenta en la siguiente.
        if imprevisto == Imprevisto.RUIDO:
            time.sleep(0.1)  # simula espera prudencial
            continue

        caminar_tramo(estado)
        atender_perro(estado)

        estado.bateria -= random.uniform(1.5, 3.0)  # simula consumo de batería

        if necesita_regresar_ya(estado):
            print("[DECISIÓN] Tiempo restante ajustado: hay que iniciar el regreso.")
            break


# ============================================================
# PASO 5: FINALIZAR EL PASEO
# ============================================================

def finalizar_paseo(estado: EstadoRobot):
    print("\n=== PASO 5: FINALIZAR EL PASEO ===")

    seguir_ruta_mas_corta_a_casa(estado)

    # Si quedó una bolsita sin depositar, se gestiona al llegar a casa
    if estado.bolsita_pendiente:
        alertar("Bolsita pendiente sin papelera cercana: se desecha en casa.")
        estado.bolsita_pendiente = False

    entregar_perro()
    guardar_registro(estado)
    avisar_dueño("El paseo ha finalizado, el perro ya está en casa.")
    estado.dueño_avisado_fin = True


# ============================================================
# FLUJO PRINCIPAL
# ============================================================

def main():
    estado = EstadoRobot()

    if not preparar_paseo(estado):
        print("\n[FIN] El paseo no se inició.")
        return

    caminar_ruta(estado)
    finalizar_paseo(estado)

    print("\n=== RESUMEN FINAL ===")
    print(f"Batería restante: {estado.bateria:.1f}%")
    print(f"Incidentes: {estado.incidentes}")


if __name__ == "__main__":
    main()
```

```python
""" MERGE WITH INTERACCION BASE
Robot Paseador de Perros - VERSIÓN INTERACTIVA
================================================
En esta versión, el programa PREGUNTA al usuario (simulando las lecturas de
los sensores del robot en tiempo real) y toma decisiones en base a esas
respuestas, mostrando por consola cada acción/decisión (outputs).

Se mantiene la misma lógica y prioridades que la versión simulada:
    1. Preparar el paseo      (inputs iniciales: batería, insumos, duración)
    2. Caminar la ruta        (bucle: 1 input round por cada tramo)
    3. Atender al perro       (inputs sobre heces / orina / sed)
    4. Resolver imprevistos   (inputs sobre ruido / lluvia / obstáculo)
    5. Finalizar el paseo     (outputs de cierre)
"""

from dataclasses import dataclass, field
from enum import Enum


# ============================================================
# CONSTANTES
# ============================================================

BATERIA_MINIMA = 20
BATERIA_CRITICA = 10
MINUTOS_POR_TRAMO_ESTIMADO = 3   # usado solo para estimar el tiempo de regreso


class Imprevisto(Enum):
    NINGUNO = "NINGUNO"
    BATERIA_BAJA = "BATERIA_BAJA"
    LLUVIA = "LLUVIA"
    OBSTACULO = "OBSTACULO"
    RUIDO = "RUIDO"
    SIN_RUTA_ALTERNATIVA = "SIN_RUTA_ALTERNATIVA"


# ============================================================
# ESTADO DEL ROBOT
# ============================================================

@dataclass
class EstadoRobot:
    bateria: float = 100.0
    total_tramos: int = 8
    tramo_actual: int = 0
    duracion_maxima_min: int = 30
    tiempo_transcurrido_min: float = 0.0

    en_casa: bool = True
    bolsita_pendiente: bool = False
    incidentes: list = field(default_factory=list)


# ============================================================
# UTILIDADES DE INPUT (validación)
# ============================================================

def pedir_si_no(mensaje: str) -> bool:
    while True:
        respuesta = input(f"{mensaje} (s/n): ").strip().lower()
        if respuesta in ("s", "si", "sí"):
            return True
        if respuesta in ("n", "no"):
            return False
        print("   → Respuesta no válida. Escribe 's' o 'n'.")


def pedir_numero(mensaje: str, minimo: float = 0, maximo: float = 100) -> float:
    while True:
        try:
            valor = float(input(f"{mensaje}: ").strip())
            if minimo <= valor <= maximo:
                return valor
            print(f"   → Ingresa un valor entre {minimo} y {maximo}.")
        except ValueError:
            print("   → Ingresa un número válido.")


def pedir_entero(mensaje: str, minimo: int = 1, maximo: int = 100) -> int:
    while True:
        try:
            valor = int(input(f"{mensaje}: ").strip())
            if minimo <= valor <= maximo:
                return valor
            print(f"   → Ingresa un número entero entre {minimo} y {maximo}.")
        except ValueError:
            print("   → Ingresa un número entero válido.")


# ============================================================
# PASO 1: PREPARAR EL PASEO (inputs iniciales)
# ============================================================

def preparar_paseo() -> EstadoRobot:
    print("\n=== PASO 1: PREPARAR EL PASEO ===")
    estado = EstadoRobot()

    estado.bateria = pedir_numero("Nivel de batería actual (%)", 0, 100)
    if estado.bateria < BATERIA_MINIMA:
        print(f"[SALIDA] Batería insuficiente ({estado.bateria}%). No se inicia el paseo.")
        estado.en_casa = True
        return None

    tiene_correa = pedir_si_no("¿Tiene correa?")
    tiene_bolsas = pedir_si_no("¿Tiene bolsas?")
    tiene_agua = pedir_si_no("¿Tiene botella de agua?")

    faltantes = []
    if not tiene_correa:
        faltantes.append("correa")
    if not tiene_bolsas:
        faltantes.append("bolsas")
    if not tiene_agua:
        faltantes.append("agua")

    if faltantes:
        print(f"[SALIDA] Falta material: {', '.join(faltantes)}. No se inicia el paseo.")
        return None

    estado.duracion_maxima_min = pedir_entero("Duración máxima del paseo (minutos)", 5, 120)
    estado.total_tramos = pedir_entero("Cantidad de tramos que tendrá la ruta", 2, 20)

    estado.en_casa = False
    print("[SALIDA] Todo listo. Saliendo de casa...")
    print("[SALIDA] Notificación al dueño: 'El paseo ha comenzado.'")
    return estado


# ============================================================
# PASO 4: RESOLVER IMPREVISTOS (inputs por tramo)
# Prioridad: batería > ruido > lluvia > obstáculo
# ============================================================

def revisar_imprevistos(estado: EstadoRobot) -> Imprevisto:
    if estado.bateria < BATERIA_CRITICA:
        print(f"[SALIDA] Batería crítica ({estado.bateria}%): se fuerza el regreso.")
        estado.incidentes.append(f"Tramo {estado.tramo_actual}: batería crítica")
        return Imprevisto.BATERIA_BAJA

    if pedir_si_no("¿Hay ruido fuerte o un perro ladrando cerca?"):
        print("[SALIDA] Robot detenido por precaución.")
        estado.incidentes.append(f"Tramo {estado.tramo_actual}: ruido fuerte, robot detenido")
        return Imprevisto.RUIDO

    if pedir_si_no("¿Está lloviendo?"):
        if pedir_si_no("¿Hay una ruta alternativa disponible?"):
            print("[SALIDA] Recalculando ruta para evitar la lluvia...")
            estado.incidentes.append(f"Tramo {estado.tramo_actual}: lluvia, ruta recalculada")
            return Imprevisto.LLUVIA
        else:
            print("[SALIDA] Lluvia sin ruta alternativa: se fuerza el regreso.")
            estado.incidentes.append(f"Tramo {estado.tramo_actual}: lluvia sin ruta alternativa")
            return Imprevisto.SIN_RUTA_ALTERNATIVA

    if pedir_si_no("¿Hay un charco u obstáculo en el camino?"):
        if pedir_si_no("¿Hay una ruta alternativa disponible?"):
            print("[SALIDA] Recalculando ruta para evitar el obstáculo...")
            estado.incidentes.append(f"Tramo {estado.tramo_actual}: obstáculo, ruta recalculada")
            return Imprevisto.OBSTACULO
        else:
            print("[SALIDA] Obstáculo sin ruta alternativa: se fuerza el regreso.")
            estado.incidentes.append(f"Tramo {estado.tramo_actual}: obstáculo sin ruta alternativa")
            return Imprevisto.SIN_RUTA_ALTERNATIVA

    return Imprevisto.NINGUNO


# ============================================================
# PASO 3: ATENDER AL PERRO (inputs por tramo)
# ============================================================

def atender_perro(estado: EstadoRobot):
    if pedir_si_no("¿El perro hizo sus necesidades (heces)?"):
        print("[SALIDA] Deteniendo... posicionando brazo... tomando bolsita... recogiendo residuos.")
        estado.bolsita_pendiente = True

    if estado.bolsita_pendiente:
        if pedir_si_no("¿Hay una papelera cerca para depositar la bolsita?"):
            print("[SALIDA] Bolsita depositada en la papelera.")
            estado.bolsita_pendiente = False
        else:
            print("[SALIDA] No hay papelera cerca: se guarda la bolsita para el siguiente tramo.")

    if pedir_si_no("¿El perro orinó?"):
        print("[SALIDA] Rociando agua para limpiar (según reglamento).")

    if pedir_si_no("¿El perro pide agua para beber?"):
        print("[SALIDA] Ofreciendo agua al perro.")


# ============================================================
# PASO 2: CAMINAR LA RUTA (bucle principal interactivo)
# ============================================================

def necesita_regresar_ya(estado: EstadoRobot) -> bool:
    tiempo_restante = estado.duracion_maxima_min - estado.tiempo_transcurrido_min
    tramos_para_volver = estado.tramo_actual
    tiempo_estimado_regreso = tramos_para_volver * MINUTOS_POR_TRAMO_ESTIMADO
    return tiempo_restante <= tiempo_estimado_regreso


def caminar_ruta(estado: EstadoRobot):
    print("\n=== PASO 2: CAMINAR LA RUTA ===")

    while not estado.en_casa and estado.tramo_actual < estado.total_tramos:
        estado.tramo_actual += 1
        print(f"\n--- Tramo {estado.tramo_actual}/{estado.total_tramos} "
              f"(batería: {estado.bateria:.1f}%, tiempo transcurrido: {estado.tiempo_transcurrido_min:.1f} min) ---")

        imprevisto = revisar_imprevistos(estado)

        if imprevisto in (Imprevisto.BATERIA_BAJA, Imprevisto.SIN_RUTA_ALTERNATIVA):
            print(f"[DECISIÓN] Imprevisto crítico ({imprevisto.value}): se interrumpe el paseo y se regresa.")
            break

        if imprevisto == Imprevisto.RUIDO:
            print("[DECISIÓN] Se espera antes de continuar; no se avanza ni se atiende al perro este tramo.")
            estado.tramo_actual -= 1  # el tramo no se contó como avance real
            continue

        print("[SALIDA] Caminando al ritmo del perro...")
        atender_perro(estado)

        estado.bateria = pedir_numero("Nivel de batería tras este tramo (%)", 0, 100)
        estado.tiempo_transcurrido_min += MINUTOS_POR_TRAMO_ESTIMADO

        if necesita_regresar_ya(estado):
            print("[DECISIÓN] El tiempo restante ya no alcanza para seguir avanzando: hay que iniciar el regreso.")
            break


# ============================================================
# PASO 5: FINALIZAR EL PASEO
# ============================================================

def finalizar_paseo(estado: EstadoRobot):
    print("\n=== PASO 5: FINALIZAR EL PASEO ===")
    print("[SALIDA] Regresando a casa por el camino más corto...")
    estado.en_casa = True

    if estado.bolsita_pendiente:
        print("[SALIDA] Bolsita pendiente sin papelera cercana: se desecha al llegar a casa.")
        estado.bolsita_pendiente = False

    print("[SALIDA] Perro entregado en casa.")
    print(f"[REGISTRO] Duración total: {estado.tiempo_transcurrido_min:.1f} min. "
          f"Batería final: {estado.bateria:.1f}%.")
    if estado.incidentes:
        print("[REGISTRO] Incidentes durante el paseo:")
        for incidente in estado.incidentes:
            print(f"   - {incidente}")
    else:
        print("[REGISTRO] Sin incidentes.")
    print("[SALIDA] Notificación al dueño: 'El paseo ha finalizado, el perro ya está en casa.'")


# ============================================================
# FLUJO PRINCIPAL
# ============================================================

def main():
    print("###########################################")
    print("#   SIMULADOR INTERACTIVO: ROBOT PASEADOR  #")
    print("###########################################")
    print("Responde cada pregunta como si fueras el sensor del robot en ese momento.\n")

    estado = preparar_paseo()
    if estado is None:
        print("\n[FIN] El paseo no se inició.")
        return

    caminar_ruta(estado)
    finalizar_paseo(estado)

    print("\n=== RESUMEN FINAL ===")
    print(f"Tramos completados: {estado.tramo_actual}/{estado.total_tramos}")
    print(f"Batería restante: {estado.bateria:.1f}%")
    print(f"Incidentes: {len(estado.incidentes)}")


if __name__ == "__main__":
    main()
```

```python
"""
Robot Paseador de Perros - VERSION INTERACTIVA CON UI MEJORADA
================================================================
Mejoras sobre la version anterior:
    - Iconos/emojis para cada tipo de evento y accion.
    - Barra visual de bateria.
    - Mini-mapa de texto con la posicion actual marcada sobre la ruta.
    - La bateria YA NO es un input manual: se descuenta automaticamente
      segun las acciones e interacciones que ocurren en cada tramo
      (caminar consume una base; recalcular ruta, recoger heces, limpiar
      orina o detenerse por ruido consumen bateria extra).
    - Los inputs se reservan solo para las decisiones que sí requieren
      "ojos humanos": imprevistos del entorno y necesidades del perro.

Estructura (igual que las versiones anteriores):
    1. Preparar el paseo
    2. Caminar la ruta
    3. Atender al perro
    4. Resolver imprevistos (transversal a 2 y 3)
    5. Finalizar el paseo
"""

from dataclasses import dataclass, field
from enum import Enum


# ============================================================
# ICONOS
# ============================================================

ICONO = {
    "casa": "🏠",
    "ubicacion": "📍",
    "bateria": "🔋",
    "perro": "🐕",
    "heces": "💩",
    "orina": "💦",
    "agua": "🚰",
    "lluvia": "🌧️",
    "obstaculo": "🚧",
    "ruido": "🔊",
    "ruta": "🗺️",
    "tiempo": "⏱️",
    "alerta": "⚠️",
    "ok": "✅",
    "falta": "❌",
    "bolsita": "📦",
    "papelera": "🗑️",
    "correa": "🦮",
    "robot": "🤖",
    "dueño": "📲",
    "registro": "📋",
}


# ============================================================
# CONSTANTES
# ============================================================

BATERIA_MINIMA = 20
BATERIA_CRITICA = 10

CONSUMO_BASE_TRAMO = 4          # % que consume caminar un tramo
CONSUMO_RECALCULO = 2           # % extra por recalcular ruta (lluvia/obstaculo)
CONSUMO_DETENERSE_RUIDO = 1     # % extra por detenerse ante ruido
CONSUMO_RECOGER_HECES = 1       # % extra por el brazo robotico recogiendo
CONSUMO_LIMPIEZA_ORINA = 1      # % extra por rociar agua
CONSUMO_OFRECER_AGUA = 0.5      # % extra por dispensar agua para beber

MINUTOS_POR_TRAMO_ESTIMADO = 3  # usado para estimar tiempo de regreso

# Ruta de ejemplo en Madrid Centro. "Casa" = Puerta del Sol.
RUTA_MADRID_CENTRO = [
    "Puerta del Sol",
    "Calle del Arenal",
    "Plaza de Isabel II (Teatro Real)",
    "Plaza de Oriente",
    "Calle Mayor",
    "Plaza Mayor",
    "Calle de Postas",
    "Puerta del Sol",
]


class Imprevisto(Enum):
    NINGUNO = "NINGUNO"
    BATERIA_BAJA = "BATERIA_BAJA"
    LLUVIA = "LLUVIA"
    OBSTACULO = "OBSTACULO"
    RUIDO = "RUIDO"
    SIN_RUTA_ALTERNATIVA = "SIN_RUTA_ALTERNATIVA"


# ============================================================
# ESTADO DEL ROBOT
# ============================================================

@dataclass
class EstadoRobot:
    bateria: float = 100.0
    ruta: list = field(default_factory=list)
    total_tramos: int = 8
    tramo_actual: int = 0
    duracion_maxima_min: int = 30
    tiempo_transcurrido_min: float = 0.0

    en_casa: bool = True
    bolsita_pendiente: bool = False
    incidentes: list = field(default_factory=list)


# ============================================================
# UTILIDADES DE INPUT (validacion)
# ============================================================

def pedir_si_no(mensaje: str) -> bool:
    while True:
        respuesta = input(f"   {mensaje} (s/n): ").strip().lower()
        if respuesta in ("s", "si", "sí"):
            return True
        if respuesta in ("n", "no"):
            return False
        print(f"   {ICONO['falta']} Respuesta no valida. Escribe 's' o 'n'.")


def pedir_numero(mensaje: str, minimo: float = 0, maximo: float = 100) -> float:
    while True:
        try:
            valor = float(input(f"{mensaje}: ").strip())
            if minimo <= valor <= maximo:
                return valor
            print(f"   {ICONO['falta']} Ingresa un valor entre {minimo} y {maximo}.")
        except ValueError:
            print(f"   {ICONO['falta']} Ingresa un numero valido.")


def pedir_entero(mensaje: str, minimo: int = 1, maximo: int = 100) -> int:
    while True:
        try:
            valor = int(input(f"{mensaje}: ").strip())
            if minimo <= valor <= maximo:
                return valor
            print(f"   {ICONO['falta']} Ingresa un numero entero entre {minimo} y {maximo}.")
        except ValueError:
            print(f"   {ICONO['falta']} Ingresa un numero entero valido.")


# ============================================================
# UI VISUAL: barra de bateria y mini-mapa
# ============================================================

def barra_bateria(nivel: float) -> str:
    nivel = max(0, min(100, nivel))
    llenos = round(nivel / 10)
    barra = "█" * llenos + "░" * (10 - llenos)
    if nivel < BATERIA_CRITICA:
        estado_txt = "CRITICA"
    elif nivel < BATERIA_MINIMA:
        estado_txt = "BAJA"
    else:
        estado_txt = "OK"
    return f"{ICONO['bateria']} [{barra}] {nivel:.1f}% ({estado_txt})"


def mostrar_mapa(estado: EstadoRobot):
    puntos = []
    for i, lugar in enumerate(estado.ruta):
        if i == estado.tramo_actual:
            puntos.append(f"{ICONO['ubicacion']}{lugar}")
        elif i == 0 or i == len(estado.ruta) - 1:
            puntos.append(f"{ICONO['casa']}{lugar}")
        else:
            puntos.append(lugar)
    print(f"   {ICONO['ruta']} " + " → ".join(puntos))


def mostrar_estado_barra(estado: EstadoRobot):
    print(f"   {barra_bateria(estado.bateria)}   "
          f"{ICONO['tiempo']} {estado.tiempo_transcurrido_min:.1f}/{estado.duracion_maxima_min} min")


# ============================================================
# PASO 1: PREPARAR EL PASEO
# ============================================================

def preparar_paseo() -> EstadoRobot:
    print(f"\n{'='*55}")
    print(f"{ICONO['robot']}  PASO 1: PREPARAR EL PASEO")
    print(f"{'='*55}")
    estado = EstadoRobot()

    estado.bateria = pedir_numero(f"{ICONO['bateria']} Nivel de bateria actual (%)", 0, 100)
    print(f"   {barra_bateria(estado.bateria)}")
    if estado.bateria < BATERIA_MINIMA:
        print(f"   {ICONO['alerta']} Bateria insuficiente. No se inicia el paseo.")
        return None

    tiene_correa = pedir_si_no(f"{ICONO['correa']} ¿Tiene correa?")
    tiene_bolsas = pedir_si_no(f"{ICONO['bolsita']} ¿Tiene bolsas?")
    tiene_agua = pedir_si_no(f"{ICONO['agua']} ¿Tiene botella de agua?")

    faltantes = []
    if not tiene_correa:
        faltantes.append("correa")
    if not tiene_bolsas:
        faltantes.append("bolsas")
    if not tiene_agua:
        faltantes.append("agua")

    if faltantes:
        print(f"   {ICONO['falta']} Falta material: {', '.join(faltantes)}. No se inicia el paseo.")
        return None

    estado.duracion_maxima_min = pedir_entero(f"{ICONO['tiempo']} Duracion maxima del paseo (minutos)", 5, 120)

    estado.ruta = RUTA_MADRID_CENTRO
    estado.total_tramos = len(estado.ruta) - 1

    print(f"\n   {ICONO['ok']} Material completo. {ICONO['ruta']} Ruta planificada "
          f"(Madrid Centro, domicilio: Puerta del Sol):")
    mostrar_mapa(estado)

    estado.en_casa = False
    print(f"   {ICONO['robot']} Saliendo de casa...")
    print(f"   {ICONO['dueño']} Notificacion al dueño: 'El paseo ha comenzado.'")
    return estado


# ============================================================
# PASO 4: RESOLVER IMPREVISTOS
# Prioridad: bateria > ruido > lluvia > obstaculo
# ============================================================

def revisar_imprevistos(estado: EstadoRobot) -> Imprevisto:
    if estado.bateria < BATERIA_CRITICA:
        print(f"   {ICONO['alerta']} Bateria critica ({estado.bateria:.1f}%): se fuerza el regreso.")
        estado.incidentes.append(f"Tramo {estado.tramo_actual}: bateria critica")
        return Imprevisto.BATERIA_BAJA

    if pedir_si_no(f"{ICONO['ruido']} ¿Hay ruido fuerte o un perro ladrando cerca?"):
        print(f"   {ICONO['robot']} Robot detenido por precaucion.")
        estado.bateria -= CONSUMO_DETENERSE_RUIDO
        estado.incidentes.append(f"Tramo {estado.tramo_actual}: ruido fuerte, robot detenido")
        return Imprevisto.RUIDO

    if pedir_si_no(f"{ICONO['lluvia']} ¿Esta lloviendo?"):
        if pedir_si_no(f"{ICONO['ruta']} ¿Hay una ruta alternativa disponible?"):
            print(f"   {ICONO['ruta']} Recalculando ruta para evitar la lluvia...")
            estado.bateria -= CONSUMO_RECALCULO
            estado.incidentes.append(f"Tramo {estado.tramo_actual}: lluvia, ruta recalculada")
            return Imprevisto.LLUVIA
        else:
            print(f"   {ICONO['alerta']} Lluvia sin ruta alternativa: se fuerza el regreso.")
            estado.incidentes.append(f"Tramo {estado.tramo_actual}: lluvia sin ruta alternativa")
            return Imprevisto.SIN_RUTA_ALTERNATIVA

    if pedir_si_no(f"{ICONO['obstaculo']} ¿Hay un charco u obstaculo en el camino?"):
        if pedir_si_no(f"{ICONO['ruta']} ¿Hay una ruta alternativa disponible?"):
            print(f"   {ICONO['ruta']} Recalculando ruta para evitar el obstaculo...")
            estado.bateria -= CONSUMO_RECALCULO
            estado.incidentes.append(f"Tramo {estado.tramo_actual}: obstaculo, ruta recalculada")
            return Imprevisto.OBSTACULO
        else:
            print(f"   {ICONO['alerta']} Obstaculo sin ruta alternativa: se fuerza el regreso.")
            estado.incidentes.append(f"Tramo {estado.tramo_actual}: obstaculo sin ruta alternativa")
            return Imprevisto.SIN_RUTA_ALTERNATIVA

    return Imprevisto.NINGUNO


# ============================================================
# PASO 3: ATENDER AL PERRO
# ============================================================

def atender_perro(estado: EstadoRobot):
    if pedir_si_no(f"{ICONO['heces']} ¿El perro hizo su caca?"):
        print(f"   {ICONO['robot']} Deteniendo... posicionando brazo... recogiendo residuos.")
        estado.bateria -= CONSUMO_RECOGER_HECES
        estado.bolsita_pendiente = True

    if estado.bolsita_pendiente:
        if pedir_si_no(f"{ICONO['papelera']} ¿Hay una papelera cerca para depositar la bolsita?"):
            print(f"   {ICONO['ok']} Bolsita depositada en la papelera.")
            estado.bolsita_pendiente = False
        else:
            print(f"   {ICONO['bolsita']} Sin papelera cerca: se guarda la bolsita para el siguiente tramo.")

    if pedir_si_no(f"{ICONO['orina']} ¿El perro orino?"):
        print(f"   {ICONO['agua']} Rociando agua para limpiar (segun reglamento).")
        estado.bateria -= CONSUMO_LIMPIEZA_ORINA

    if pedir_si_no(f"{ICONO['perro']} ¿El perro pide agua para beber?"):
        print(f"   {ICONO['agua']} Ofreciendo agua al perro.")
        estado.bateria -= CONSUMO_OFRECER_AGUA


# ============================================================
# PASO 2: CAMINAR LA RUTA (bucle principal)
# ============================================================

def necesita_regresar_ya(estado: EstadoRobot) -> bool:
    tiempo_restante = estado.duracion_maxima_min - estado.tiempo_transcurrido_min
    tramos_para_volver = estado.tramo_actual
    tiempo_estimado_regreso = tramos_para_volver * MINUTOS_POR_TRAMO_ESTIMADO
    return tiempo_restante <= tiempo_estimado_regreso


def caminar_ruta(estado: EstadoRobot):
    print(f"\n{'='*55}")
    print(f"{ICONO['robot']}  PASO 2: CAMINAR LA RUTA")
    print(f"{'='*55}")

    while not estado.en_casa and estado.tramo_actual < estado.total_tramos:
        origen = estado.ruta[estado.tramo_actual]
        destino = estado.ruta[estado.tramo_actual + 1]

        print(f"\n--- {ICONO['ubicacion']} Tramo {estado.tramo_actual + 1}/{estado.total_tramos}: "
              f"de {origen} hacia {destino} ---")
        mostrar_estado_barra(estado)

        imprevisto = revisar_imprevistos(estado)

        if imprevisto in (Imprevisto.BATERIA_BAJA, Imprevisto.SIN_RUTA_ALTERNATIVA):
            print(f"   {ICONO['alerta']} Imprevisto critico ({imprevisto.value}): "
                  f"se interrumpe el paseo y se regresa.")
            break

        if imprevisto == Imprevisto.RUIDO:
            print(f"   {ICONO['robot']} Se espera antes de continuar; no se avanza este tramo.")
            continue

        estado.tramo_actual += 1
        print(f"   {ICONO['perro']} Caminando al ritmo del perro hacia {destino}...")
        atender_perro(estado)

        estado.bateria -= CONSUMO_BASE_TRAMO
        estado.tiempo_transcurrido_min += MINUTOS_POR_TRAMO_ESTIMADO

        print(f"   {barra_bateria(estado.bateria)}")

        if necesita_regresar_ya(estado):
            print(f"   {ICONO['alerta']} El tiempo restante ya no alcanza para seguir: hay que regresar.")
            break


# ============================================================
# PASO 5: FINALIZAR EL PASEO
# ============================================================

def finalizar_paseo(estado: EstadoRobot):
    print(f"\n{'='*55}")
    print(f"{ICONO['robot']}  PASO 5: FINALIZAR EL PASEO")
    print(f"{'='*55}")
    print(f"   {ICONO['casa']} Regresando a Puerta del Sol por el camino mas corto...")
    estado.en_casa = True

    if estado.bolsita_pendiente:
        print(f"   {ICONO['bolsita']} Bolsita pendiente sin papelera cercana: se desecha en casa.")
        estado.bolsita_pendiente = False

    print(f"   {ICONO['ok']} Perro entregado en casa.")
    print(f"\n   {ICONO['registro']} REGISTRO DEL PASEO")
    print(f"   {ICONO['tiempo']} Duracion total: {estado.tiempo_transcurrido_min:.1f} min")
    print(f"   {barra_bateria(estado.bateria)}")
    if estado.incidentes:
        print(f"   {ICONO['alerta']} Incidentes:")
        for incidente in estado.incidentes:
            print(f"      - {incidente}")
    else:
        print(f"   {ICONO['ok']} Sin incidentes.")
    print(f"   {ICONO['dueño']} Notificacion al dueño: 'El paseo ha finalizado, el perro ya esta en casa.'")


# ============================================================
# FLUJO PRINCIPAL
# ============================================================

def main():
    print("┌" + "─"*53 + "┐")
    print(f"│  {ICONO['robot']}  SIMULADOR INTERACTIVO: ROBOT PASEADOR          │"[:55] + "│")
    print("└" + "─"*53 + "┘")
    print("Responde cada pregunta como si fueras el sensor del robot en ese momento.")
    print(f"({ICONO['bateria']} La bateria ya NO se pregunta: se descuenta sola segun lo que pase en el paseo)\n")

    estado = preparar_paseo()
    if estado is None:
        print(f"\n{ICONO['falta']} El paseo no se inicio.")
        return

    caminar_ruta(estado)
    finalizar_paseo(estado)

    print(f"\n{'='*55}")
    print(f"{ICONO['registro']}  RESUMEN FINAL")
    print(f"{'='*55}")
    print(f"   Tramos completados: {estado.tramo_actual}/{estado.total_tramos}")
    print(f"   {barra_bateria(estado.bateria)}")
    print(f"   Incidentes: {len(estado.incidentes)}")


if __name__ == "__main__":
    main()
```
